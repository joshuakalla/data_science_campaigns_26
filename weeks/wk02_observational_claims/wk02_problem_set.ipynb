Every paper and vendor memo you will read reports comparisons as a **regression** rather than a `groupby`. Here is the vendor's gap in that form. Run the cell below. You do **not** need to write any regression code yourself.

```python
import statsmodels.formula.api as smf

# Regress 2022 turnout on whether the district ran the program.
model = smf.ols('turnout_2022 ~ ran_program', data=districts).fit()
model.params
```
