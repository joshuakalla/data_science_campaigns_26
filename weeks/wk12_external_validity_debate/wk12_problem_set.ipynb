1. **W1:** Conflating fixed and marginal costs.
2. **W2:** Presenting an observational comparison as a causal effect.
3. **W3:** Applying a result from one setting to a different one.

5. **W5:** Treating an underpowered null as evidence of no effect.
6. **W7:** Reporting the complier comparison as a causal effect.

9. **W10:** Trusting a model's probability without its calibration record.
10. **W11:** Tipping-point math written to justify a decision already made.

## Task 1: Read a table

This is the course's terminal skill, and Exam 2 asks you to do it again on a table you have not seen.

Below is a regression table from a (real) GOTV canvassing-mailer experiment, run only in **low-salience** elections. Read it and answer the questions. **No code — just read it, the way you'd read a table in a paper.**

**Outcome: `voted_2014` (1 = voted, 0 = did not)**

| Variable | Coefficient | Std. error |
|---|---|---|
| Intercept | 0.614\*\*\* | 0.004 |
| **treatment (got the mailer)** | **0.0104\*** | **0.004** |
| vote history: average | −0.206\*\*\* | 0.005 |
| vote history: below median | −0.481\*\*\* | 0.005 |

*N = 48,426  ·  R² = 0.19  ·  \*p \< 0.05, \*\*\*p \< 0.001*

**Note the format.** Published tables usually do **not** print a p-value column — they mark significance with
asterisks and give you a legend, exactly as above. That is what you are reading here, and it is what the exam
hands you. `.summary()` in Colab prints the p-value instead; a journal table makes you read the stars.

**Questions — one sentence each.** Answer them in the cell below. If you run out of time, submit what you have; this is graded on effort, not completeness.

**1a.** What is the estimated effect of the mailer on turnout, in percentage points? (Remember: ×100 turns a coefficient on a 0/1 outcome into points.)

**1b.** Is the treatment effect statistically significant at the 0.05 level? **How can you tell from the table, given that there is no p-value column?** Is it significant at the 0.001 level?

**1c.** The 95% confidence interval for the treatment coefficient is about [0.003, 0.018] — that is, **[+0.3, +1.8] percentage points**. In plain English, what does that interval tell the donor?

**1d.** What do the two "vote history" rows mean, and why are their coefficients so much bigger than the treatment coefficient? Does that make the mailer look unimportant?

**1e.** This table is from **low-salience** elections only, and the effect (+1.0 pp) is bigger than the +0.7 pp from the big 17-state midterm study. Which number should the donor use for a 2026 *midterm general*, and why? (This is the external-validity judgment.)

**1f.** The R² is 0.19. Does a low R² mean the mailer's estimated effect is wrong, or unreliable? (Careful — these are two different questions, and R² answers neither of them directly. What *does* tell you whether to trust the estimate?)
