## 3. The two tactics from today's case

Let's encode each tactic as variables and compute its average cost per unit delivered.

### Mailer

- Fixed cost: \$0 (creative already paid)
- Marginal cost: \$0.50 per household

```python
mailer_fixed_cost = 0
mailer_marginal_cost = 0.50

mailer_households = (budget - mailer_fixed_cost) / mailer_marginal_cost
mailer_average_cost_per_household = budget / mailer_households

print('Mailer: households reached =', mailer_households)
print('Mailer: average cost per household = $', mailer_average_cost_per_household)
```

### Digital ad

- Fixed cost: \$30,000 (production)
- Marginal cost: \$0.05 per impression

```python
digital_fixed_cost = 30000
digital_marginal_cost = 0.05

digital_impressions = (budget - digital_fixed_cost) / digital_marginal_cost
digital_average_cost_per_impression = budget / digital_impressions

print('Digital ad: impressions delivered =', digital_impressions)
print('Digital ad: average cost per impression = $', digital_average_cost_per_impression)
```

The digital ad's average cost per impression (\$0.125) is **2.5× its marginal cost** (\$0.05). The production cost is spread over only 400,000 impressions.

## 4. What if the budget were bigger?

Let's compare three scales: \$50K, \$500K, and \$5M.

```python
# What if the budget were $50,000? $500,000? $5,000,000?
# We compute the average cost per unit for each tactic, three times.

print('At $50,000:')
print('  mailer  average cost per household =  $', round(50000 / ((50000 - mailer_fixed_cost) / mailer_marginal_cost), 4))
print('  digital average cost per impression = $', round(50000 / ((50000 - digital_fixed_cost) / digital_marginal_cost), 4))

print('At $500,000:')
print('  mailer  average cost per household =  $', round(500000 / ((500000 - mailer_fixed_cost) / mailer_marginal_cost), 4))
print('  digital average cost per impression = $', round(500000 / ((500000 - digital_fixed_cost) / digital_marginal_cost), 4))

print('At $5,000,000:')
print('  mailer  average cost per household =  $', round(5000000 / ((5000000 - mailer_fixed_cost) / mailer_marginal_cost), 4))
print('  digital average cost per impression = $', round(5000000 / ((5000000 - digital_fixed_cost) / digital_marginal_cost), 4))
```

At \$50K the digital ad's average cost is **2.5×** its marginal cost (\$0.125 vs. \$0.05). At \$500K it's down to \$0.053. At \$5M it is essentially equal to the marginal cost. Tactics with high fixed costs become more attractive as you scale them up.
