## Task 1: Cost per vote (pre-filled)

**cost per vote = cost / (effect size as a proportion)**

One catch, and it is the whole point of the first cell. \$20 buys one canvassing **conversation**, but +2.0 pp is the effect of one **door knocked**, and most doors do not open. Week 5: canvassers reach about 30% of the doors they knock, so a door costs \$20 × 0.30 = \$6. Phone is the same: \$2.00 a completed call at a 25% contact rate is \$0.50 a dial.

The `contact_rate` column holds those numbers. Mail and texts are 1.00: you pay to send, and everyone on the list gets one. **Run the cell.**

```python
# Cost of one attempt, then cost per additional vote
tactics['cost_per_attempt'] = tactics['cost_per_contact'] * tactics['contact_rate']
tactics['cost_per_vote'] = tactics['cost_per_attempt'] / (tactics['effect_pp'] / 100)
tactics[['tactic', 'contact_rate', 'cost_per_attempt', 'cost_per_vote']].round(2)
```
