# Week 13 Sanity-Check: The \$30-per-Vote Claim

The cost memo (Artifact 4) says the program costs **\$30 per additional vote** — cheaper than canvassing or mail. Before you believe any cost-per-vote, **re-derive it.** A cost-per-vote is just:

`cost / (voters reached x effect size)`

so it depends entirely on two choices: *which effect size* and *which costs*. Let's see how much those two choices move the answer.

```python
voters   = 200_000     # voters the $2M program would reach (the memo's number)
marginal = 300_000     # per-contact cost (texts, platform, volunteer support)
fixed    = 500_000     # staff, training, building the app -- the memo calls this "overhead"

def cost_per_vote(effect_pct, include_fixed):
    votes = voters * (effect_pct / 100)
    cost  = marginal + (fixed if include_fixed else 0)
    return cost / votes
```

## The memo's number vs. an honest one

The memo uses the **+5.0%** from the *county comparison* (Artifact 2 — confounded) and counts **marginal cost only**.

An honest version uses the *experiment's* defensible effect. The RCT's pre-registered primary outcome was small and non-significant — **they never report the exact number**, so we plug in a *plausible* \~**+0.7%** to see what it would imply (the +3.2% was a cherry-picked subgroup, not the primary). And the \$500K of fixed cost is a real cost of running the program.

```python
print(f"MEMO claim   (+5.0%, marginal only):  ${cost_per_vote(5.0, False):.0f} per vote")
print(f"HONEST version (+0.7%, all costs):      ${cost_per_vote(0.7, True):.0f} per vote")
```

## How much do the two choices matter? (a sensitivity table)

Same program, same \$2M. The cost-per-vote swings from \$30 to \~\$570 purely from which effect size and which costs you pick.

```python
effects = [5.0, 3.2, 0.7]   # county comparison / cherry-picked subgroup / RCT primary outcome
print(f"{'effect':>8} | {'marginal only':>14} | {'all costs':>12}")
print('-'*40)
for e in effects:
    print(f"{e:>7}% | ${cost_per_vote(e, False):>12,.0f} | ${cost_per_vote(e, True):>10,.0f}")
```

**Read the table.** The memo lives in the top-left corner (\$30). The defensible number is the bottom-right (**\~\$571** — *worse* than the \~\$80 mail and \~\$100 canvassing it claims to beat). The other cells show it's the *assumptions*, not the program, doing the work. Even the *most charitable* honest cell (+3.2% subgroup, all costs) is \~\$125 — still above mail.
